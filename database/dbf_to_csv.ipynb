```python
INPUT_FOLDER = "Data BPS/dbf"   # e.g. "C:/data/dbf_files" or "/home/user/dbf_files"
OUTPUT_FOLDER = "Data BPS/csv"  # e.g. "C:/data/csv_output" or "/home/user/csv_output"
```

```python
os.makedirs(OUTPUT_FOLDER, exist_ok=True)

dbf_files = [f for f in os.listdir(INPUT_FOLDER) if f.lower().endswith(".dbf")]

for filename in dbf_files:
    dbf_path = os.path.join(INPUT_FOLDER, filename)
    csv_path = os.path.join(
        OUTPUT_FOLDER,
        os.path.splitext(filename)[0] + ".csv"
    )

    table = dbf.Table(dbf_path)
    table.open()

    with open(csv_path, "w", newline="", encoding="utf-8-sig") as f:
        writer = csv.writer(f)

        # Header
        writer.writerow(table.field_names)

        # Records
        for record in table:
            writer.writerow([record[field] for field in table.field_names])

    table.close()

    print(f"Converted {filename} ({len(table)} rows)")

print("Done.")
```
